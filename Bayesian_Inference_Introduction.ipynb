# Overview

Bayesian inference can be thought of as a theory of **learning under uncertainty**.

In many situations, we do not know the true state of the world. We may not know the value of a parameter, the correct model, the hidden cause of an observation, or the future outcome of a process. However, we are rarely in a state of complete ignorance. We usually begin with some prior information: previous experience, physical constraints, expert knowledge, symmetry assumptions, or even a deliberately vague state of belief.

Bayesian inference formalizes this situation by saying:

> Knowledge is not all-or-nothing; it comes in degrees of belief.

These degrees of belief are represented by probability measures.

The unknown quantity, say $\theta$, is treated as uncertain. This does **not necessarily mean** that $\theta$ is physically random. It may be fixed but unknown. The probability distribution over $\theta$ represents our state of knowledge about it.

Before observing data, we have a **prior belief** represented by a probability measure
$\Pi$
on a measurable parameter space $(\Theta,\mathfrak T)$.


After an observation, the data does not usually reveal $\theta$ exactly. Rather, it changes which values of $\theta$ are more or less plausible. Bayesian inference updates the prior belief $\Pi$ into a **posterior belief.**


So the philosophical idea of  Bayesian inference is:

> Bayesian inference is rational belief revision in the presence of evidence.



This is one of the key differences between Bayesian and purely frequentist thinking. In a frequentist view, the parameter $\theta$ is fixed and probability describes the long-run behavior of repeated data. In a Bayesian view, probability may also describe uncertainty about $\theta$ itself, conditional on the information currently available.

Thus, Bayesian inference is not merely a computational rule. It is a coherent framework for answering the question:

> Given what I believed before, and given what I have now observed, what should I believe next?

In measure-theoretic language, this becomes simply:

> Bayesian inference is conditioning a joint probability measure on the observed data.

In the following we rigorously define this notion.

# Mathematical Formulation

Let $(\Omega,\mathfrak A,P)$ be an underlying probability space. Let

$$
X:\Omega\to \mathscr X
$$

be a random variable representing the unknown parameter, latent state, or hidden quantity of interest (traditionaly it is common to use the variable $\theta$ to denote paramters. However in order to stick to the general conditioning notation adopted in this course we will use $X$ instead). Here $(\mathscr X,\mathfrak U_X)$ is a measurable space. Let

$$
Y:\Omega\to \mathscr Y
$$

be a random variable representing the observed data where $(\mathscr Y,\mathfrak U_Y)$ is another measurable space.

Thus, $X$ and $Y$ are measurable maps satisfying
$X^{-1}(A)\in\mathfrak A$ for all $A\in\mathfrak U_X$,
and $Y^{-1}(B)\in\mathfrak A$ for all $B\in\mathfrak U_Y$.
Together, they define a joint random variable

$$
Z=(X,Y):\Omega\to \mathscr X\times\mathscr Y,
$$

where the product space is equipped with the product $\sigma$-algebra $\mathfrak U_X\otimes\mathfrak U_Y$.

The joint law of $(X,Y)$ is the push-forward measure

$$
P_{X,Y}\triangleq (X,Y)_{\#} P.
$$

That is, for every measurable set $C\in \mathfrak U_X\otimes\mathfrak U_Y$,

$$
P_{X,Y}(C)=P\big((X,Y)^{-1}(C)\big).
$$

The marginal law of $X$ is

$$
P_X\triangleq X_{\#}P,
$$

and the marginal law of $Y$ is

$$
P_Y\triangleq Y_{\#}P.
$$

Explicitly,

$$
P_X(A)=P(X\in A),
\qquad A\in\mathfrak U_X,
$$

and

$$
P_Y(B)=P(Y\in B),
\qquad B\in\mathfrak U_Y.
$$

In the Bayesian interpretation, $P_X$ is the **prior law** of the unknown quantity $X$ before observing the data $Y$.



# Prior, Likelihood, and Posterior via Regular Conditional Laws



In Bayesian inference, the prior, likelihood, and posterior are not fundamentally densities. They are probability measures or conditional probability kernels. Densities appear only when suitable dominating measures exist.

Assume that regular conditional distributions exist. Then there exists, for each $x\in\mathscr X$,
a probability measure on $(\mathscr Y,\mathfrak U_Y)$, such that for each $B\in\mathfrak U_Y$, the probability that the observed data lies in $B$, conditional on the parameter value $X=x$ is given by
$$
P_{Y\mid X=x}(B).
$$


Similarly, the posterior is described by a regular probability measure on $(\mathscr X,\mathfrak U_X)$
such that for each $A\in\mathfrak U_X$, the probability that the parameter $X$ lies in $A$, conditional on the observation $Y=y$ is given by
$$
P_{X\mid Y=y}(A)
$$

This is the **posterior law of $X$ after observing $Y=y$**.



>Thus, measure-theoretically, Bayesian inference is the passage from the prior law
$$
P_X
$$
>to the posterior law
$$
P_{X\mid Y=y}
$$
after observing the data value $Y=y$.

For every $A\in\mathfrak U_X$ and every $B\in\mathfrak U_Y$,

$$
P(X\in A,Y\in B)=\int_B P_{X\mid Y=y}(A)\,dP_Y(y).
$$

Equivalently,

$$
P_{X,Y}(A\times B)=\int_B P_{X\mid Y=y}(A)\,dP_Y(y).
$$

This identity says that the joint law $P_{X,Y}$ can be disintegrated with respect to the marginal law $P_Y$ of the observed data:

$$
dP_{X,Y}(x,y)=
dP_{X\mid Y=y}(x)\,dP_Y(y).
$$

On the other hand, before observing $Y$, the Bayesian model may be described by the prior law of $X$ together with the conditional law of $Y$ given $X$. That is,
$$
dP_{X,Y}(x,y)=dP_{Y\mid X=x}(y)\,dP_X(x).
$$

Thus, the same joint probability measure admits two factorizations.

Therefore, the Bayesian update is the passage

$$
dP_X(x)
\quad\longmapsto\quad
dP_{X\mid Y=y}(x),
$$
after observing the value $Y=y$.

The measure-theoretic form of Bayes' rule is the identity
$$
dP_{Y\mid X=x}(y)\,dP_X(x)=
dP_{X\mid Y=y}(x)\,dP_Y(y).
$$
It can be written in the compact symbolic form:

$$
dP_{X\mid Y=y}(x)=
\frac{
dP_{Y\mid X=x}(y)\,dP_X(x)
}{
dP_Y(y)
}.
$$
>This expression should be interpreted carefully: it is not ordinary division of measures in two independent variables. It is just short hand for the identity preceding it.

The marginal probability of finding $Y\in B$ is given by
$$
P_{Y}(B)=\int_{\mathscr{X}} P_{Y\mid X=x}(B)\,dP_X(x).
$$

# Density Form of Bayes' Rule

Suppose now that $\mathscr X$ and $\mathscr Y$ are Euclidean spaces, and suppose that the relevant measures admit densities with respect to suitable reference measures.

Let $\lambda_X$ be a reference measure on $\mathscr X$, and let $\lambda_Y$ be a reference measure on $\mathscr Y$.

Assume that the prior law $P_X$ is absolutely continuous with respect to $\lambda_X$. Then the prior density is defined by

$$
dP_X(x)=f_X(x){d\lambda_X}(x).
$$

Assume also that, for each $x$, the conditional law $P_{Y\mid X=x}$ is absolutely continuous with respect to $\lambda_Y$. Then the likelihood density is defined by

$$
dP_{Y\mid X=x}=f_{Y\mid X}(y\mid x)d\lambda_Y(y).
$$

The **word likelihood** refers to the same function viewed as a function of $x$ after the data value $y$ has been fixed. Thus, for fixed $y$,

$$
x\mapsto f_{Y\mid X}(y\mid x)
$$

is the likelihood function.

If the joint law $P_{X,Y}$ admits a density $f_{X,Y}$ with respect to $\lambda_X\otimes\lambda_Y$, then

$$
f_{X,Y}(x,y)=f_{Y\mid X}(y\mid x)f_X(x).
$$

The marginal density of $Y$ is obtained by integrating out $x$:

$$
f_Y(y)=\int_{\mathscr X} f_{Y\mid X}(y\mid x)f_X(x)\,d\lambda_X(x).
$$

Whenever $f_Y(y)>0$, the posterior density is

$$
f_{X\mid Y}(x\mid y)=\frac{f_{X,Y}(x,y)}{f_Y(y)}.
$$

Therefore,

$$
f_{X\mid Y}(x\mid y)=\frac{
f_{Y\mid X}(y\mid x)f_X(x)
}{
\int_{\mathscr X} f_{Y\mid X}(y\mid x')f_X(x')\,d\lambda_X(x')
}.
$$

This is the usual density form of Bayes' rule.

In symbolic Bayesian notation, this is often summarized as

$$
\text{posterior}
\propto
\text{likelihood}\times \text{prior}.
$$

More explicitly,

$$
f_{X\mid Y}(x\mid y)
\propto
f_{Y\mid X}(y\mid x)f_X(x),
$$

where the constant of proportionality is determined by the condition

$$
\int_{\mathscr X} f_{X\mid Y}(x\mid y)\,d\lambda_X(x)=1.
$$



# Conditional Expectation as Bayesian Point Estimation

The posterior distribution gives the full updated uncertainty about $X$ after observing $Y=y$. If we want a single point estimate of $X$, a natural Bayesian choice is the posterior mean.

Measure-theoretically, this is described by the conditional expectation

$$
\mathbb E[X\mid Y].
$$

Assume that $X$ is integrable, i.e.

$$
\mathbb E[|X|]<\infty.
$$

Then $\mathbb E[X\mid Y]$ is the unique $\sigma(Y)$-measurable random variable satisfying

$$
\int_G \mathbb E[X\mid Y]\,dP=\int_G X,dP
\qquad
\forall G\in\sigma(Y).
$$

Since $\mathbb E[X\mid Y]$ is $\sigma(Y)$-measurable, there exists a measurable function $g:\mathscr Y\to\mathscr X$ such that

$$
\mathbb E[X\mid Y]=g(Y).
$$

The function $g$ is the Bayesian estimator associated with squared-error loss.

For a particular observation $Y=y$, the posterior mean is

$$
g(y)=\mathbb E[X\mid Y=y].
$$

In terms of the posterior measure, this is

$$
g(y)=\int_{\mathscr X} x\,dP_{X\mid Y=y}(x).
$$

If the posterior admits a density, then

$$
g(y)=\int_{\mathscr X} x\,f_{X\mid Y}(x\mid y)\,d\lambda_X(x).
$$

Thus, the Bayesian point estimate is obtained by averaging over the posterior distribution.



## Optimality of the Posterior Mean

The posterior mean is not merely a convenient estimate. It is optimal under mean-square error.

Let $Z$ be any other estimator of $X$ based on the observed data $Y$. This means that $Z$ is $\sigma(Y)$-measurable.

Then

$$
\mathbb E\left[|X-\mathbb E[X\mid Y]|^2\right]
\le
\mathbb E\left[|X-Z|^2\right].
$$

Therefore, $\mathbb E[X\mid Y]$ is the optimal estimator of $X$ among all $\sigma(Y)$-measurable estimators under squared-error loss.

Equivalently,

$$
\mathbb E[X\mid Y]=\operatorname*{arg,min}_{Z\in L^2(\sigma(Y))}
\mathbb E\left[|X-Z|^2\right].
$$

The remaining uncertainty after observing $Y$ is described by the conditional variance

$$
\operatorname{Var}(X\mid Y)=\mathbb E[X^2\mid Y]
\left(\mathbb E[X\mid Y]\right)^2
$$

in the scalar case.

For vector-valued $X$, the conditional covariance is

$$
\operatorname{Cov}(X\mid Y)=\mathbb E\left[
\left(X-\mathbb E[X\mid Y]\right)
\left(X-\mathbb E[X\mid Y]\right)^T
\mid Y
\right].
$$



# Examples

## The Multivariate Gaussian Case

A central example occurs when $X$ and $Y$ are jointly Gaussian.

Suppose

$$
\begin{bmatrix}
X\\
Y
\end{bmatrix}
\sim
\mathscr N
\left(
\begin{bmatrix}
\mu_X\\
\mu_Y
\end{bmatrix},
\begin{bmatrix}
\Sigma_{XX} & \Sigma_{XY}\\
\Sigma_{YX} & \Sigma_{YY}
\end{bmatrix}
\right).
$$

Assume that $\Sigma_{YY}$ is invertible.

Then the conditional distribution of $X$ given $Y=y$ is Gaussian:

$$
X\mid Y=y
\sim
\mathscr N
\left(
\mu_{X\mid y},
\Sigma_{X\mid y}
\right),
$$

where

$$
\mu_{X\mid y}=\mu_X
+
\Sigma_{XY}\Sigma_{YY}^{-1}(y-\mu_Y),
$$

and

$$
\Sigma_{X\mid y}=\Sigma_{XX}-
\Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}.
$$

Therefore, as a random variable,

$$
\mathbb E[X\mid Y]=\mu_X
+
\Sigma_{XY}\Sigma_{YY}^{-1}(Y-\mu_Y).
$$

The matrix

$$
\Sigma_{XY}\Sigma_{YY}^{-1}
$$

acts as a gain matrix. It determines how strongly the innovation

$$
Y-\mu_Y
$$

should change the prior mean $\mu_X$.

Thus, in the Gaussian case, Bayesian inference has a closed-form update:

$$
\text{posterior mean}=\text{prior mean}
+
\text{gain}\times\text{innovation}.
$$

This is the same mathematical structure that appears in linear Gaussian estimation and the Kalman filter.



## Example: Estimating the Bias of a Coin

Suppose we are given a coin, but we do not know whether it is fair. The coin may be biased toward heads, biased toward tails, or approximately fair. Let

$$
\Theta\in[0,1]
$$

denote the unknown probability that the coin lands heads.

The value of $\Theta$ is fixed for the coin, but it is unknown to us. Therefore, in the Bayesian formulation, we represent our uncertainty about $\Theta$ using a probability distribution on the interval $[0,1]$.

Before tossing the coin, we begin with a prior belief about possible values of $\Theta$. For example, if we have no strong reason to prefer one value over another, we may start with a uniform prior on $[0,1]$.

We then toss the coin several times and observe the number of heads and tails. These observations provide evidence about the unknown value of $\Theta$. If we observe many heads, then larger values of $\Theta$ become more plausible. If we observe many tails, then smaller values of $\Theta$ become more plausible.

Bayesian inference updates the prior distribution of $\Theta$ into a posterior distribution after seeing the data. The posterior distribution describes our revised uncertainty about the probability of heads.


Let $\Theta \sim \operatorname{Beta}(\alpha,\beta)$.

The prior density is

$$
f_\Theta(\theta)=\frac{1}{B(\alpha,\beta)}
\theta^{\alpha-1}(1-\theta)^{\beta-1},
\qquad 0<\theta<1.
$$


Now suppose we toss the coin $n$ times and observe $h$ heads and $t=n-h$ tails. The likelihood is

$$P(Y=(h,t)\mid \Theta=\theta)=\binom{n}{h}\theta^h(1-\theta)^t.$$


The posterior density is proportional to

$$
f_{\Theta\mid Y}(\theta\mid Y)
\propto
P(Y=(h,t)\mid \Theta=\theta)f_\Theta(\theta).
$$

Therefore,

$$
f_{\Theta\mid Y}(\theta\mid h,t)
\propto
\theta^h(1-\theta)^t
\theta^{\alpha-1}(1-\theta)^{\beta-1}.
$$

So

$$
f_{\Theta\mid Y}(\theta\mid h,t)
\propto
\theta^{\alpha+h-1}(1-\theta)^{\beta+t-1}.
$$

Hence,

$$
\Theta\mid Y=(h,t)
\sim
\operatorname{Beta}(\alpha+h,\beta+t).
$$

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

def plot_beta_prior_posterior(n, h, alpha=1, beta_param=1, num_points=1000):
    """
    Plot the prior and posterior distributions for the unknown coin bias theta.

    Parameters
    ----------
    n : int
        Total number of coin tosses.
    h : int
        Number of heads observed.
    alpha : float
        Prior Beta distribution first parameter.
    beta_param : float
        Prior Beta distribution second parameter.
    num_points : int
        Number of grid points for plotting.
    """

    if not isinstance(n, int) or not isinstance(h, int):
        raise ValueError("n and h must be integers.")

    if n < 0:
        raise ValueError("n must be non-negative.")

    if h < 0 or h > n:
        raise ValueError("h must satisfy 0 <= h <= n.")

    if alpha <= 0 or beta_param <= 0:
        raise ValueError("alpha and beta_param must be positive.")

    t = n - h

    posterior_alpha = alpha + h
    posterior_beta = beta_param + t

    theta = np.linspace(0, 1, num_points)

    prior_density = beta.pdf(theta, alpha, beta_param)
    posterior_density = beta.pdf(theta, posterior_alpha, posterior_beta)

    prior_mean = alpha / (alpha + beta_param)
    posterior_mean = posterior_alpha / (posterior_alpha + posterior_beta)

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=theta,
            y=prior_density,
            mode="lines",
            name=f"Prior: Beta({alpha}, {beta_param})"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=theta,
            y=posterior_density,
            mode="lines",
            name=f"Posterior: Beta({posterior_alpha}, {posterior_beta})"
        )
    )

    fig.add_vline(
        x=prior_mean,
        line_dash="dash",
        annotation_text=f"Prior mean = {prior_mean:.3f}",
        annotation_position="top left"
    )

    fig.add_vline(
        x=posterior_mean,
        line_dash="dash",
        annotation_text=f"Posterior mean = {posterior_mean:.3f}",
        annotation_position="top right"
    )

    fig.update_layout(
        title=f"Bayesian Update for Coin Bias: n={n}, heads={h}, tails={t}",
        xaxis_title="Coin bias θ = P(Heads)",
        yaxis_title="Density",
        template="plotly_white",
        hovermode="x unified"
    )

    fig.show()

    print(f"Prior:      Beta({alpha}, {beta_param})")
    print(f"Posterior: Beta({posterior_alpha}, {posterior_beta})")
    print(f"Prior mean:      {prior_mean:.4f}")
    print(f"Posterior mean: {posterior_mean:.4f}")
```

```python
plot_beta_prior_posterior(n=10, h=8)
```

## Example: Estimating Loan Default Probability


Suppose a bank issues many similar personal loans. Each borrower either defaults or does not default within one year.

Let $\Theta\in[0,1]$
denote the unknown probability that a randomly selected borrower defaults within one year.
The bank does not know the true value of $\Theta$, but it may have prior information from past years, industry knowledge, or expert judgment.

Before observing the current portfolio, suppose the bank believes that the default probability is usually around $5\%$. We can encode this using a Beta prior, for example

$$
\Theta\sim \operatorname{Beta}(5,95).
$$

This prior has mean

$$
\mathbb E[\Theta]=\frac{5}{5+95}=0.05.
$$

So the prior belief says:

$$
\text{default probability is probably around }5\%.
$$

Now suppose the bank observes $n=200$ loans from the current year, and $d=16$ of them default. Conditional on $\Theta=\theta$, the number of defaults is modeled as

$$
D\mid \Theta=\theta \sim \operatorname{Binomial}(n,\theta).
$$

The likelihood is

$$
P(D=d\mid \Theta=\theta)=\binom{n}{d}\theta^d(1-\theta)^{n-d}.
$$

The Bayesian update is

$$
\text{posterior}
\propto
\text{likelihood}\times \text{prior}.
$$

Since the prior is Beta and the likelihood is Binomial, the posterior is again Beta:

$$
\Theta\mid D=d
\sim
\operatorname{Beta}(\alpha+d,\beta+n-d).
$$

For our example,

$$
\Theta\sim \operatorname{Beta}(5,95),
$$

and we observe

$$
n=200,
\qquad
d=16.
$$

Therefore,

$$
\Theta\mid D=16
\sim
\operatorname{Beta}(5+16,;95+184).
$$

Hence,

$$
\Theta\mid D=16
\sim
\operatorname{Beta}(21,279).
$$

The posterior mean is

$$
\mathbb E[\Theta\mid D=16]=\frac{21}{21+279}=0.07.
$$

So the bank's estimate of the default probability moves from $5\%$
to $7\%$.

The interpretation is:

$$
\boxed{
\text{The observed data suggests default risk is higher than initially believed.}
}
$$



We can describe the update as

$$
\operatorname{Beta}(5,95)
\quad\longrightarrow\quad
\operatorname{Beta}(21,279).
$$
This example is useful because it mirrors a real financial risk problem:

$$
\text{How risky is this loan portfolio?}
$$

The prior comes from historical knowledge. The likelihood comes from the observed default count. The posterior gives the updated belief about the true default probability.

In words:

$$
\boxed{
\text{Bayesian inference updates the bank's belief about default risk after observing actual loan outcomes.}
}
$$

Graphically, this is very nice. The prior density is centered near $5\%$. After observing $16$ defaults out of $200$, the posterior density shifts to the right and becomes narrower, centered near $7\%$.

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

def plot_default_probability_update(
    n=200,
    d=16,
    alpha=5,
    beta_param=95,
    num_points=1000
):
    """
    Visualize Bayesian updating for a loan default probability.

    Model
    -----
    Theta = unknown default probability

    Prior:
        Theta ~ Beta(alpha, beta_param)

    Data:
        D | Theta = theta ~ Binomial(n, theta)

    Posterior:
        Theta | D=d ~ Beta(alpha + d, beta_param + n - d)

    Parameters
    ----------
    n : int
        Number of loans observed.
    d : int
        Number of defaults observed.
    alpha : float
        First parameter of the Beta prior.
    beta_param : float
        Second parameter of the Beta prior.
    num_points : int
        Number of grid points for plotting.
    """

    if not isinstance(n, int) or not isinstance(d, int):
        raise ValueError("n and d must be integers.")

    if n <= 0:
        raise ValueError("n must be positive.")

    if d < 0 or d > n:
        raise ValueError("d must satisfy 0 <= d <= n.")

    if alpha <= 0 or beta_param <= 0:
        raise ValueError("alpha and beta_param must be positive.")

    non_defaults = n - d

    posterior_alpha = alpha + d
    posterior_beta = beta_param + non_defaults

    theta = np.linspace(0, 0.20, num_points)

    prior_density = beta.pdf(theta, alpha, beta_param)
    posterior_density = beta.pdf(theta, posterior_alpha, posterior_beta)

    prior_mean = alpha / (alpha + beta_param)
    posterior_mean = posterior_alpha / (posterior_alpha + posterior_beta)

    observed_default_rate = d / n

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=theta,
            y=prior_density,
            mode="lines",
            name=f"Prior: Beta({alpha}, {beta_param})"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=theta,
            y=posterior_density,
            mode="lines",
            name=f"Posterior: Beta({posterior_alpha}, {posterior_beta})"
        )
    )

    fig.add_vline(
        x=prior_mean,
        line_dash="dash",
        annotation_text=f"Prior mean = {prior_mean:.2%}",
        annotation_position="top left"
    )

    fig.add_vline(
        x=posterior_mean,
        line_dash="dash",
        annotation_text=f"Posterior mean = {posterior_mean:.2%}",
        annotation_position="top right"
    )

    fig.add_vline(
        x=observed_default_rate,
        line_dash="dot",
        annotation_text=f"Observed rate = {observed_default_rate:.2%}",
        annotation_position="bottom right"
    )

    fig.update_layout(
        title=f"Bayesian Update of Loan Default Probability: {d} Defaults out of {n} Loans",
        xaxis_title="Default probability θ",
        yaxis_title="Density",
        template="plotly_white",
        hovermode="x unified",
        legend=dict(
            x=0.70,
            y=0.95,
            bgcolor="rgba(255,255,255,0.7)"
        )
    )

    fig.update_xaxes(tickformat=".0%")

    fig.show()

    print("Bayesian default probability update")
    print("-----------------------------------")
    print(f"Observed loans:       {n}")
    print(f"Defaults observed:    {d}")
    print(f"Non-defaults:         {non_defaults}")
    print()
    print(f"Prior:                Beta({alpha}, {beta_param})")
    print(f"Posterior:            Beta({posterior_alpha}, {posterior_beta})")
    print()
    print(f"Prior mean:           {prior_mean:.4f}  = {prior_mean:.2%}")
    print(f"Observed default rate:{observed_default_rate:.4f}  = {observed_default_rate:.2%}")
    print(f"Posterior mean:       {posterior_mean:.4f}  = {posterior_mean:.2%}")
```

```python
plot_default_probability_update(
    n=200,
    d=16,
    alpha=5,
    beta_param=95
)
```

## 2PL Example: Bayesian Estimation of a User Ability Parameter from Item Responses

An online learning platform presents a user with a set of $n$ multiple-choice questions. Each question is either answered correctly or incorrectly.

Let

$$
Y_i=
\begin{cases}
1, & \text{if the user answers item } i \text{ correctly},\\
0, & \text{if the user answers item } i \text{ incorrectly}.
\end{cases}
$$

The platform assumes that the probability of a correct response is governed by a two-parameter logistic item response model. Specifically, conditional on the user's latent ability parameter $\Theta=\theta$, the response probability for item $i$ is

$$
P(Y_i=1\mid \Theta=\theta)=\frac{1}{1+e^{-a_i(\theta-b_i)}},
$$

where $a_i>0$ is the discrimination parameter of item $i$, and $b_i$ is the difficulty parameter of item $i$.

The item parameters

$$
(a_i,b_i),\qquad i=1,\dots,n,
$$

are assumed to be known from previous calibration. The unknown quantity is the user's latent ability

$$
\Theta\in\mathbb R.
$$

Before observing the user's responses, suppose the platform assigns the prior distribution

$$
\Theta\sim \mathscr N(0,1).
$$

The user then answers (n) items, producing an observed response vector

$$
y=(y_1,\dots,y_n)\in{0,1}^n.
$$

Assuming that the item responses are conditionally independent given $\Theta=\theta$, the likelihood is

$$
P(Y=y\mid \Theta=\theta)=\prod_{i=1}^n
\left[
p_i(\theta)
\right]^{y_i}
\left[
1-p_i(\theta)
\right]^{1-y_i},
$$

where

$$
p_i(\theta)
=\frac{1}{1+\exp{-a_i(\theta-b_i)}}.
$$

Therefore, the posterior density of $\Theta$ given the observed response vector $Y=y$ is

$$
f_{\Theta\mid Y}(\theta\mid y)=\frac{
\left[
\prod_{i=1}^n
p_i(\theta)^{y_i}
(1-p_i(\theta))^{1-y_i}
\right]
f_\Theta(\theta)
}{
\int_{\mathbb R}
\left[
\prod_{i=1}^n
p_i(s)^{y_i}
(1-p_i(s))^{1-y_i}
\right]
f_\Theta(s),ds
}.
$$

Here,

$$
f_\Theta(\theta)=\frac{1}{\sqrt{2\pi}}
\exp\left(-\frac{\theta^2}{2}\right)
$$

is the prior density.

The Bayesian estimate of the user's ability under squared-error loss is the posterior mean

$$
\widehat{\theta}_{\mathrm{Bayes}}
=\mathbb E[\Theta\mid Y=y]
=\int_{\mathbb R}
\theta f_{\Theta\mid Y}(\theta\mid y),d\theta.
$$

Alternatively, the maximum a posteriori estimate is

$$
\widehat{\theta}_{\mathrm{MAP}}
\in
\operatorname{arg,max}_{\theta\in\mathbb R}
f_{\Theta\mid Y}(\theta\mid y).
$$

## Example: Online Bayesian Estimation of the State of the Economy


An important problem in macroeconomics is determining the current state of the
economy from economic data observed over time. In particular, policymakers,
businesses, and economists may wish to determine whether the economy is currently
in an \emph{expansion} or a \emph{recession}.

The true economic regime is not directly observed. Instead, economic indicators
such as industrial production, employment, income, and sales provide noisy
information about the underlying state of the economy.

Let the latent economic regime during month $k$ be represented by the random
variable

$$
S_k \in \{0,1\},
$$

where

$$
S_k =
\begin{cases}
0, & \text{the economy is in expansion},\\
1, & \text{the economy is in recession}.
\end{cases}
$$

The objective is to estimate recursively the probability

$$
\pi_k
\triangleq
P(S_k=1\mid Y_1=y_1,\ldots,Y_k=y_k),
$$

which represents our belief, based on all economic information observed up to
month $k$, that the economy is currently in recession.

---

#### Evolution of the Economic Regime

Economic regimes tend to persist. If the economy is currently expanding, it is
likely to remain in expansion during the following month. Similarly, recessions
typically persist for several months.

We model this using a two-state Markov process. Define

$$
p_{ij}
=
P(S_k=j\mid S_{k-1}=i),
\qquad i,j\in\{0,1\}.
$$

The transition probability matrix is therefore

$$
P=
\begin{bmatrix}
p_{00} & p_{01}\\
p_{10} & p_{11}
\end{bmatrix},
$$

where

$$
p_{00}+p_{01}=1,
\qquad
p_{10}+p_{11}=1.
$$

For example, $p_{01}$ represents the probability that an economy currently in
expansion enters recession during the next month, while $p_{11}$ represents the
probability that an economy currently in recession remains in recession.

---

#### Economic Observations

Suppose that at each month we observe an economic indicator $Y_k$. For example,
$Y_k$ may represent the monthly growth rate of industrial production.

The statistical behavior of the indicator depends on the underlying economic
regime. Assume

$$
Y_k\mid S_k=0
\sim
\mathscr N(\mu_0,\sigma_0^2)
$$

during expansions, and

$$
Y_k\mid S_k=1
\sim
\mathscr N(\mu_1,\sigma_1^2)
$$

during recessions.

Typically one would expect

$$
\mu_1 < \mu_0,
$$

since measures of economic activity tend to grow more slowly, or decline, during
recessions.

Let

$$
f_s(y)
\triangleq
f_{Y_k\mid S_k}(y\mid s),
\qquad s\in\{0,1\},
$$

denote the corresponding conditional probability density.

Hence

$$
f_s(y)
=
\frac{1}{\sqrt{2\pi\sigma_s^2}}
\exp\left[
-\frac{(y-\mu_s)^2}{2\sigma_s^2}
\right].
$$

The parameters

$$
p_{01},\quad p_{11},\quad
\mu_0,\quad\mu_1,\quad
\sigma_0^2,\quad\sigma_1^2
$$

are assumed to have been estimated previously from historical economic data and
are treated as known in the online estimation problem.

---

#### Bayesian Prediction

Suppose that after observing the data up to month $k-1$ we know

$$
\pi_{k-1}
=
P(S_{k-1}=1\mid y_{1:k-1}),
$$

where

$$
y_{1:k-1}
=
(y_1,\ldots,y_{k-1}).
$$

Consequently,

$$
P(S_{k-1}=0\mid y_{1:k-1})
=
1-\pi_{k-1}.
$$

Before observing the economic indicator for month $k$, use the Markov transition
model and the law of total probability to determine the predicted probability

---

$$
\pi_k^-
\triangleq
P(S_k=1\mid y_{1:k-1}).
$$


This represents the prior probability that the economy is in recession immediately
before observing the new economic data.


Let

$$
\pi_k^-
\triangleq
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1}),
$$

where

$$
Y_{1:k-1}=(Y_1,\ldots,Y_{k-1}).
$$

The events $\{S_{k-1}=0\}$ and $\{S_{k-1}=1\}$ form a partition.
Therefore, by the conditional law of total probability,

$$
\begin{aligned}
\pi_k^-
&=
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1})
\\[2mm]
&=
\sum_{i\in\{0,1\}}
P(S_k=1\mid S_{k-1}=i,Y_{1:k-1}=y_{1:k-1})
\\
&\hspace{25mm}\times
P(S_{k-1}=i\mid Y_{1:k-1}=y_{1:k-1}).
\end{aligned}
$$

Expanding the sum gives

$$
\begin{aligned}
\pi_k^-
&=
P(S_k=1\mid S_{k-1}=0,Y_{1:k-1}=y_{1:k-1})
\\
&\hspace{10mm}\times
P(S_{k-1}=0\mid Y_{1:k-1}=y_{1:k-1})
\\[1mm]
&\quad+
P(S_k=1\mid S_{k-1}=1,Y_{1:k-1}=y_{1:k-1})
\\
&\hspace{10mm}\times
P(S_{k-1}=1\mid Y_{1:k-1}=y_{1:k-1}).
\end{aligned}
$$

Because $\{S_k\}$ is a first-order Markov process, its next state
depends on the past only through $S_{k-1}$. Hence,

$$
P(S_k=1\mid S_{k-1}=i,Y_{1:k-1}=y_{1:k-1})
=
P(S_k=1\mid S_{k-1}=i).
$$

From the definition of the transition probabilities,

$$
P(S_k=1\mid S_{k-1}=0)=p_{01}
$$

and

$$
P(S_k=1\mid S_{k-1}=1)=p_{11}.
$$

Furthermore, by the definition of $\pi_{k-1}$,

$$
P(S_{k-1}=1\mid Y_{1:k-1}=y_{1:k-1})
=
\pi_{k-1},
$$

and, because $S_{k-1}$ has only two possible states,

$$
P(S_{k-1}=0\mid Y_{1:k-1}=y_{1:k-1})
=
1-\pi_{k-1}.
$$

Substituting these results gives

$$
\boxed{
\pi_k^-
=
p_{01}(1-\pi_{k-1})
+
p_{11}\pi_{k-1}.
}
$$

The two terms correspond to the only two ways in which the economy
can be in recession during month $k$:

$$
\underbrace{
p_{01}(1-\pi_{k-1})
}_{\substack{
\text{expansion at }k-1\\
\text{followed by recession at }k
}}
+
\underbrace{
p_{11}\pi_{k-1}
}_{\substack{
\text{recession at }k-1\\
\text{followed by recession at }k
}}.
$$

Thus, $\pi_k^-$ is the predicted recession probability before
observing the new economic indicator $Y_k$.

---

#### Bayesian Update

At month $k$, suppose that the newly released economic indicator takes
the observed value

$$
Y_k=y_k.
$$

The objective is to determine the updated probability

$$
P(S_k=1\mid Y_{1:k}=y_{1:k}).
$$

---

Let

$$
\pi_k
\triangleq
P(S_k=1\mid Y_{1:k}=y_{1:k})
$$

denote the posterior recession probability after observing $Y_k=y_k$.
The predicted recession probability before observing $Y_k$ is

$$
\pi_k^-
\triangleq
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1}).
$$

Since $Y_k$ is continuously distributed, consider a small observation
element $dy_k$ containing $y_k$. By the definition of conditional
probability,

If the relevant conditional densities exist and are continuous at
$y_k$, define

$$
D_\varepsilon(y_k)
\triangleq
(y_k-\varepsilon,y_k+\varepsilon).
$$

Then the posterior recession probability can be written as

\begin{align*}
\pi_k
&=
P(S_k=1\mid Y_k=y_k,Y_{1:k-1}=y_{1:k-1})
\\
&=
\lim_{\varepsilon\to0}
P\!\left(
S_k=1
\mid
Y_k\in D_\varepsilon(y_k),
Y_{1:k-1}=y_{1:k-1}
\right).
\end{align*}

By the definition of conditional probability,

\begin{align*}
\pi_k
&=
\lim_{\varepsilon\to0}
\frac{
P\!\left(
S_k=1,
Y_k\in D_\varepsilon(y_k)
\mid
Y_{1:k-1}=y_{1:k-1}
\right)
}{
P\!\left(
Y_k\in D_\varepsilon(y_k)
\mid
Y_{1:k-1}=y_{1:k-1}
\right)
}.
\end{align*}

Using the conditional multiplication rule in the numerator gives

\begin{align*}
\pi_k
&=
\lim_{\varepsilon\to0}
\frac{
\begin{aligned}
&P\!\left(
Y_k\in D_\varepsilon(y_k)
\mid
S_k=1,Y_{1:k-1}=y_{1:k-1}
\right)
\\
&\qquad\times
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1})
\end{aligned}
}{
P\!\left(
Y_k\in D_\varepsilon(y_k)
\mid
Y_{1:k-1}=y_{1:k-1}
\right)
}.
\end{align*}

The conditional probabilities involving $Y_k$ can be expressed as
integrals of the corresponding conditional densities:

\begin{align*}
&P\!\left(
Y_k\in D_\varepsilon(y_k)
\mid
S_k=1,Y_{1:k-1}=y_{1:k-1}
\right)
\\
&\qquad=
\int_{D_\varepsilon(y_k)}
f_{Y_k\mid S_k,Y_{1:k-1}}
(u\mid 1,y_{1:k-1})\,du,
\end{align*}

and

\begin{align*}
&P\!\left(
Y_k\in D_\varepsilon(y_k)
\mid
Y_{1:k-1}=y_{1:k-1}
\right)
\\
&\qquad=
\int_{D_\varepsilon(y_k)}
f_{Y_k\mid Y_{1:k-1}}
(u\mid y_{1:k-1})\,du.
\end{align*}

Therefore,

\begin{align*}
\pi_k
&=
\lim_{\varepsilon\to0}
\frac{
\left[
\displaystyle
\int_{D_\varepsilon(y_k)}
f_{Y_k\mid S_k,Y_{1:k-1}}
(u\mid 1,y_{1:k-1})\,du
\right]
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1})
}{
\displaystyle
\int_{D_\varepsilon(y_k)}
f_{Y_k\mid Y_{1:k-1}}
(u\mid y_{1:k-1})\,du
}.
\end{align*}

Since the conditional densities are continuous at $y_k$,

\begin{align*}
&\lim_{\varepsilon\to0}
\frac{
\displaystyle
\int_{D_\varepsilon(y_k)}
f_{Y_k\mid S_k,Y_{1:k-1}}
(u\mid 1,y_{1:k-1})\,du
}{
2\varepsilon
}
\\
&\qquad=
f_{Y_k\mid S_k,Y_{1:k-1}}
(y_k\mid 1,y_{1:k-1}),
\end{align*}

and

\begin{align*}
&\lim_{\varepsilon\to0}
\frac{
\displaystyle
\int_{D_\varepsilon(y_k)}
f_{Y_k\mid Y_{1:k-1}}
(u\mid y_{1:k-1})\,du
}{
2\varepsilon
}
\\
&\qquad=
f_{Y_k\mid Y_{1:k-1}}
(y_k\mid y_{1:k-1}).
\end{align*}

Dividing both the numerator and denominator by $2\varepsilon$ and
taking the limit gives the density form of Bayes' rule:

\begin{align*}
\pi_k
&=
\frac{
f_{Y_k\mid S_k,Y_{1:k-1}}
(y_k\mid 1,y_{1:k-1})
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1})
}{
f_{Y_k\mid Y_{1:k-1}}
(y_k\mid y_{1:k-1})
}.
\end{align*}

The observation model assumes that, conditional on the current regime
$S_k$, the current observation $Y_k$ is independent of the previous
observations $Y_{1:k-1}$. Therefore,

$$
f_{Y_k\mid S_k,Y_{1:k-1}}
(y_k\mid s,y_{1:k-1})
=
f_{Y_k\mid S_k}(y_k\mid s)
=
f_s(y_k).
$$

Also, by definition,

$$
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1})
=
\pi_k^-,
$$

and hence

$$
P(S_k=0\mid Y_{1:k-1}=y_{1:k-1})
=
1-\pi_k^-.
$$

Thus, the numerator of the Bayesian update is

$$
f_1(y_k)\pi_k^-.
$$

To determine the denominator, apply the conditional law of total
probability over the two possible values of $S_k$:

\begin{align*}
&f_{Y_k\mid Y_{1:k-1}}
(y_k\mid y_{1:k-1})
\\
&=
f_{Y_k\mid S_k,Y_{1:k-1}}
(y_k\mid 1,y_{1:k-1})
P(S_k=1\mid Y_{1:k-1}=y_{1:k-1})
\\
&\quad+
f_{Y_k\mid S_k,Y_{1:k-1}}
(y_k\mid 0,y_{1:k-1})
P(S_k=0\mid Y_{1:k-1}=y_{1:k-1})
\\
&=
f_1(y_k)\pi_k^-
+
f_0(y_k)(1-\pi_k^-).
\end{align*}

Substituting the numerator and denominator into Bayes' rule gives

$$
\boxed{
\pi_k
=
\frac{
f_1(y_k)\pi_k^-
}{
f_1(y_k)\pi_k^-
+
f_0(y_k)(1-\pi_k^-)
}.
}
$$

The posterior probability $\pi_k$ then becomes the information used to
predict the recession probability for month $k+1$.

Thus we also have

$$
P(S_k=0\mid y_{1:k})
=
1-\pi_k.
$$

Thus, every new economic observation produces a new posterior belief about the
current state of the economy.

The posterior probability $\pi_k$ then becomes the information used to construct
the prior probability $\pi_{k+1}^-$ for the following month.

Therefore, the procedure generates the recursive Bayesian estimator

$$
\boxed{
\pi_{k-1}
\;\longrightarrow\;
\pi_k^-
\;\xrightarrow{\;Y_k=y_k\;}\;
\pi_k.
}
$$

Starting from an initial recession probability

$$
\pi_0=P(S_0=1),
$$

this recursion can be applied sequentially to an entire economic time series to
produce

$$
\pi_1,\pi_2,\ldots,\pi_N.
$$

The resulting sequence provides a real-time probabilistic estimate of the
underlying economic regime.

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

from scipy.stats import norm
from plotly.subplots import make_subplots


def run_recession_bayes_filter(
    df,
    date_col="date",
    indicator_col="Ind_Prod",
    p01=0.02,
    p11=0.85,
    mu_0=0.30,
    sigma_0=0.70,
    mu_1=-1.00,
    sigma_1=1.00,
    pi0=None,
    standardize=True,
    recession_threshold=0.50
):
    """
    Estimate recession probabilities recursively using a two-state
    Bayesian Markov filter.

    States
    ------
    S_k = 0 : expansion
    S_k = 1 : recession

    Transition probabilities
    ------------------------
    p01 = P(S_k=1 | S_{k-1}=0)
    p11 = P(S_k=1 | S_{k-1}=1)

    Observation model
    -----------------
    Y_k | S_k=0 ~ Normal(mu_0, sigma_0^2)
    Y_k | S_k=1 ~ Normal(mu_1, sigma_1^2)

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame containing dates and the economic indicator.
    date_col : str
        Name of the date column.
    indicator_col : str
        Name of the economic indicator column.
    p01, p11 : float
        Markov transition probabilities.
    mu_0, sigma_0 : float
        Expansion-state observation parameters.
    mu_1, sigma_1 : float
        Recession-state observation parameters.
    pi0 : float or None
        Initial recession probability. If None, the stationary
        recession probability of the Markov chain is used.
    standardize : bool
        If True, standardize the growth-rate series.
    recession_threshold : float
        Probability above which the regime is classified as recession.

    Returns
    -------
    results : pandas.DataFrame
        Data, predicted probabilities, likelihoods and posterior
        recession probabilities.
    fig : plotly.graph_objects.Figure
        Interactive Plotly figure.
    """

    # -------------------------------------------------------------
    # 1. Check model parameters
    # -------------------------------------------------------------
    if not (0 <= p01 <= 1):
        raise ValueError("p01 must lie between 0 and 1.")

    if not (0 <= p11 <= 1):
        raise ValueError("p11 must lie between 0 and 1.")

    if sigma_0 <= 0 or sigma_1 <= 0:
        raise ValueError("sigma_0 and sigma_1 must be positive.")

    # -------------------------------------------------------------
    # 2. Prepare the economic data
    # -------------------------------------------------------------
    data = df[[date_col, indicator_col]].copy()

    data[date_col] = pd.to_datetime(data[date_col], errors="coerce")

    data[indicator_col] = pd.to_numeric(
        data[indicator_col],
        errors="coerce"
    )

    data = (
        data
        .dropna(subset=[date_col, indicator_col])
        .sort_values(date_col)
        .drop_duplicates(subset=date_col)
        .reset_index(drop=True)
    )

    # Percentage change in industrial production
    data["indicator_growth"] = (
        100.0 * data[indicator_col].pct_change()
    )

    # Remove the first observation and any invalid growth values
    data = data.replace([np.inf, -np.inf], np.nan)
    data = data.dropna(subset=["indicator_growth"]).reset_index(drop=True)

    if len(data) == 0:
        raise ValueError(
            "No usable observations remain after computing growth."
        )

    # -------------------------------------------------------------
    # 3. Construct the observation sequence Y_k
    # -------------------------------------------------------------
    if standardize:
        growth_mean = data["indicator_growth"].mean()
        growth_std = data["indicator_growth"].std(ddof=0)

        if growth_std <= 0:
            raise ValueError(
                "The indicator growth series has zero variance."
            )

        data["Y"] = (
            data["indicator_growth"] - growth_mean
        ) / growth_std

        observation_label = "Standardized industrial-production growth"
    else:
        data["Y"] = data["indicator_growth"]
        observation_label = "Industrial-production growth (%)"

    # -------------------------------------------------------------
    # 4. Set the initial recession probability
    # -------------------------------------------------------------
    p10 = 1.0 - p11

    if pi0 is None:
        # Stationary recession probability:
        # pi = p01 / (p01 + p10)
        if p01 + p10 > 0:
            pi0 = p01 / (p01 + p10)
        else:
            pi0 = 0.5

    if not (0 <= pi0 <= 1):
        raise ValueError("pi0 must lie between 0 and 1.")

    # -------------------------------------------------------------
    # 5. Recursive Bayesian estimator
    # -------------------------------------------------------------
    pi_previous = pi0

    predicted_probabilities = []
    expansion_likelihoods = []
    recession_likelihoods = []
    posterior_probabilities = []

    for y_k in data["Y"].to_numpy():

        # Prediction:
        #
        # pi_k^- = p01(1-pi_{k-1}) + p11 pi_{k-1}
        pi_predicted = (
            p01 * (1.0 - pi_previous)
            + p11 * pi_previous
        )

        # Observation likelihoods
        f0 = norm.pdf(y_k, loc=mu_0, scale=sigma_0)
        f1 = norm.pdf(y_k, loc=mu_1, scale=sigma_1)

        # Bayesian denominator / predictive density
        denominator = (
            f1 * pi_predicted
            + f0 * (1.0 - pi_predicted)
        )

        if denominator <= 0:
            raise FloatingPointError(
                "The Bayesian normalizing constant became zero."
            )

        # Update:
        #
        # pi_k =
        # f1(y_k) pi_k^- /
        # [f1(y_k) pi_k^- + f0(y_k)(1-pi_k^-)]
        pi_updated = (
            f1 * pi_predicted / denominator
        )

        predicted_probabilities.append(pi_predicted)
        expansion_likelihoods.append(f0)
        recession_likelihoods.append(f1)
        posterior_probabilities.append(pi_updated)

        # The current posterior becomes the previous posterior
        # for the next iteration
        pi_previous = pi_updated

    data["pi_predicted"] = predicted_probabilities
    data["f_expansion"] = expansion_likelihoods
    data["f_recession"] = recession_likelihoods
    data["pi_recession"] = posterior_probabilities

    data["estimated_regime"] = np.where(
        data["pi_recession"] >= recession_threshold,
        "Recession",
        "Expansion"
    )

    # -------------------------------------------------------------
    # 6. Plot the results
    # -------------------------------------------------------------
    fig = make_subplots(
        rows=2,
        cols=1,
        shared_xaxes=True,
        vertical_spacing=0.10,
        row_heights=[0.45, 0.55],
        subplot_titles=(
            observation_label,
            "Online Bayesian recession probability"
        )
    )

    fig.add_trace(
        go.Scatter(
            x=data[date_col],
            y=data["Y"],
            mode="lines",
            name=observation_label,
            line=dict(color="#1f77b4", width=1.5),
            customdata=np.column_stack([
                data["indicator_growth"]
            ]),
            hovertemplate=(
                "Date: %{x|%Y-%m-%d}<br>"
                "Observation: %{y:.3f}<br>"
                "Growth: %{customdata[0]:.3f}%"
                "<extra></extra>"
            )
        ),
        row=1,
        col=1
    )

    fig.add_hline(
        y=0,
        line_dash="dot",
        line_color="gray",
        row=1,
        col=1
    )

    fig.add_trace(
        go.Scatter(
            x=data[date_col],
            y=data["pi_predicted"],
            mode="lines",
            name="Predicted probability",
            line=dict(
                color="#ff7f0e",
                width=1.5,
                dash="dash"
            ),
            hovertemplate=(
                "Date: %{x|%Y-%m-%d}<br>"
                "Predicted probability: %{y:.3f}"
                "<extra></extra>"
            )
        ),
        row=2,
        col=1
    )

    fig.add_trace(
        go.Scatter(
            x=data[date_col],
            y=data["pi_recession"],
            mode="lines",
            name="Posterior recession probability",
            line=dict(color="#d62728", width=2.5),
            fill="tozeroy",
            fillcolor="rgba(214,39,40,0.15)",
            customdata=np.column_stack([
                data["Y"],
                data["f_expansion"],
                data["f_recession"]
            ]),
            hovertemplate=(
                "Date: %{x|%Y-%m-%d}<br>"
                "Posterior probability: %{y:.3f}<br>"
                "Observation: %{customdata[0]:.3f}<br>"
                "Expansion likelihood: %{customdata[1]:.4f}<br>"
                "Recession likelihood: %{customdata[2]:.4f}"
                "<extra></extra>"
            )
        ),
        row=2,
        col=1
    )

    fig.add_hline(
        y=recession_threshold,
        line_dash="dash",
        line_color="black",
        annotation_text=(
            f"Classification threshold = {recession_threshold:.2f}"
        ),
        annotation_position="top left",
        row=2,
        col=1
    )

    fig.update_yaxes(
        title_text="Observation",
        row=1,
        col=1
    )

    fig.update_yaxes(
        title_text="Probability",
        range=[0, 1],
        tickformat=".0%",
        row=2,
        col=1
    )

    fig.update_xaxes(
        title_text="Date",
        row=2,
        col=1
    )

    fig.update_layout(
        title=(
            "Online Bayesian Estimation of the U.S. Economic Regime"
        ),
        template="plotly_white",
        height=750,
        hovermode="x unified",
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="left",
            x=0
        )
    )


    nber_recessions = [
        ("1973-11-01", "1975-03-01"),
        ("1980-01-01", "1980-07-01"),
        ("1981-07-01", "1982-11-01"),
        ("1990-07-01", "1991-03-01"),
        ("2001-03-01", "2001-11-01"),
        ("2007-12-01", "2009-06-01"),
        ("2020-02-01", "2020-04-01")
    ]

    for start, end in nber_recessions:
        fig.add_vrect(
            x0=start,
            x1=end,
            fillcolor="gray",
            opacity=0.20,
            line_width=0,
            layer="below",
            row="all",
            col=1
        )

    fig.update_layout(
        title=(
            "Online Bayesian Recession Probability "
            "with Official NBER Recession Periods"
        )
    )

    fig.show()


    return data, fig
```

# PyMC

> **PyMC** is a probabilistic programming library in Python for constructing and fitting Bayesian statistical models. It allows the user to specify prior distributions, likelihood models, deterministic relationships, and observed data in a readable Python syntax that closely mirrors the mathematical structure of a Bayesian model. Once the model is specified, PyMC uses computational inference algorithms such as Markov chain Monte Carlo and variational inference to approximate the posterior distribution of the unknown quantities. In this sense, PyMC turns the abstract Bayesian update from prior to posterior into a practical computational procedure.

In the language of your measure-theoretic formulation, PyMC helps us implement the passage

$$
dP_X(x)
\quad\longmapsto\quad dP_{X\mid Y=y}(x),
$$

when this posterior measure is too difficult to compute analytically. It does not change the Bayesian idea; it provides machinery for carrying it out.
